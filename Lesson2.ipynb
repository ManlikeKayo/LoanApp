```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python
all_models = {
    'Logistics Regression':(model, xtest),
    'RandomForestClassifier':(rf_model, xtest),
    'DecisionTreeClassifier':(dt_model, xtest),
    
}
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, (all_models, res)) in zip(axes, all_models.items()):
    y_prediction = model.predict(res)
    cm = confusion_matrix(ytest, y_prediction)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Rejected (0)', 'Approve (1)'])
    disp.plot(ax=ax, cmap='Greens')
```
